```python
# Part 1: String Concatenation

str1 = "Hello"
name = input("Enter your Name: ")

# Concatenate "Hello" and name
greeting = str1 + " " + name
print(greeting)

# Concatenate welcome message
welcome_msg = ", welcome to Python programming"
final_str = greeting + welcome_msg
print(final_str)
```

```python
# Part 2: String Slicing and Indexing

print("a. First character:", final_str[0])
print("b. Last character:", final_str[-1])
print("c. First 5 characters:", final_str[:5])
print("d. Last 11 characters:", final_str[-11:])
print("e. Reversed string:", final_str[::-1])

# Extract the word "Python"
start_idx = final_str.find("Python")
print("f. Extracted word:", final_str[start_idx : start_idx + 6])
```

```python
# Part 3: String Methods

strM = "Python beginner tutorial"

print("a. Uppercase:", strM.upper())
print("b. Lowercase:", strM.lower())
print("c. Capitalized:", strM.capitalize())
print("d. Count of 't':", strM.count('t'))
print("e. Replaced string:", strM.replace("Python", "Machine Learning"))
```

```python
# Part 4: Tuples

tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)

# Concatenate
t_combine = tuple1 + tuple2
print("a. t_combine:", t_combine)

# Repeat 3 times
print("b. Repeated 3 times:", t_combine * 3)

# Access elements
print("c. 3rd element:", t_combine[2])
print("d. First three elements:", t_combine[:3])
print("e. Last three elements:", t_combine[-3:])
```
