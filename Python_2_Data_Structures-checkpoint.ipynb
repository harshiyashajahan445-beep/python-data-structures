```python
# 1. List creation
age_list = [24, 25, 26, 27, 28]
name_list = ["Alice", "Bob", "Charlie", "David", "Emma"]

print("Initial age_list:", age_list)
print("Initial name_list:", name_list)
```

```python
# 2. List Operations / Modifications
# a. Append the string "Yazhini" to name_list
name_list.append("Yazhini")

# b. Insert the element 30 at index 2 in age_list
age_list.insert(2, 30)

# c. Remove the string "Yazhini" from name_list
name_list.remove("Yazhini")

# d. Pop the last element from age_list
popped_element = age_list.pop()
print("Popped element from age_list:", popped_element)

# e. Extend the age_list with additional ages [29, 30, 26]
age_list.extend([29, 30, 26])

# f. Sort age_list in descending order
age_list.sort(reverse=True)

# g. Find Max age, Min age and sum of all ages from age_list
max_age = max(age_list)
min_age = min(age_list)
sum_ages = sum(age_list)

print("Modified age_list:", age_list)
print("Max age:", max_age)
print("Min age:", min_age)
print("Sum of all ages:", sum_ages)
```

```python
# 3. Accessing List Elements
# a. Print the first element of name_list
print("First element:", name_list[0])

# b. Print the last element of name_list
print("Last element:", name_list[-1])

# c. Print the elements from index 2 to index 4 in name_list
print("Elements from index 2 to 4:", name_list[2:5])

# d. Print the elements of name_list in reverse order
print("Reversed name_list:", name_list[::-1])
```

```python
# Performance Category Program
# 1. Prompt user for Input (Score range from 0 to 10 inclusive)[cite: 2]
try:
    score_input = input("Enter your score (0 to 10): ")
    score = float(score_input) # using float/int to handle numeric checks
    
    if 0 <= score <= 10:
        # 2 & 4. Find performance category and assign customized prompt
        if score > 7:
            print(f"Above Average: Excellent job! Your score of {score} shows stellar mastery and consistent hard work.")
        elif 4 <= score <= 7:
            print(f"Average: Good effort! Keep practicing, there's room for improvement[cite: 2]. (Score: {score})")
        else:
            print(f"Below Average: Need to Improve your performance, consistent practice will lead to better results[cite: 2]. (Score: {score})")
    else:
        print("Invalid input! Please enter a score strictly between 0 and 10.")
except ValueError:
    print("Invalid input! Please enter a valid numerical value.")
```
